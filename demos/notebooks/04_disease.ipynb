# Demo 04: Disease Investigation**Story**: "Alien diseases: detect, diagnose, and cure."Measure a healthy baseline, apply a disease, detect symptoms, and evaluate diagnostic agents.

```python
import sys, ossys.path.insert(0, os.path.join(os.getcwd(), "demos"))from _shared import make_disease_system, oracle_agent, random_agentfrom alienbio.bio import AgentInterface, DiagnoseTask, detect_symptomsfrom alienbio.viz import symptom_chart, concentration_trajectory
```

## Healthy Baseline

```python
system, baseline, perturbations = make_disease_system(seed=42)for mol, conc in baseline.steady_state.items():    print(f"{mol}: {conc:.4f}")
```

## Apply Disease

```python
disease = perturbations[0]print(f"Disease: {disease.name} ({disease.kind} on {disease.target_reaction})")disease.apply(system)diseased_tl = system.run(steps=300)
```

## Detect Symptoms

```python
concentrations = {name: system.state[name] for name in system.chemistry.molecules}symptoms = detect_symptoms(concentrations, baseline)for s in symptoms:    direction = "HIGH" if s.value > s.healthy_range.high else "LOW"    print(f"{s.molecule}: {s.value:.4f} ({direction}, deviation={s.deviation:.4f})")
```

```python
symptom_chart(symptoms, baseline, title="Symptom Chart")
```

## Diagnosis

```python
task = DiagnoseTask(perturbations, applied_index=0)iface = AgentInterface(system)oracle_result = task.score(iface, oracle_agent(iface, task))random_result = task.score(iface, random_agent(iface, task))print(f"Oracle: {oracle_result.score:.2f}")print(f"Random: {random_result.score:.2f}")
```

```python
concentration_trajectory(diseased_tl, title="Diseased Trajectories")
```

**Takeaway**: Disease = structural perturbation; the task framework evaluates agent reasoning.
