# Demo 05: Multi-Compartment Organisms**Story**: "Real organisms have structure."Generate a multi-organ organism and visualize transport dynamics.

```python
import sys, ossys.path.insert(0, os.path.join(os.getcwd(), "demos"))from _shared import make_organismfrom alienbio.bio import WorldStateImplfrom alienbio.viz import compartment_heatmap
```

## Generate Organism

```python
organism, chem = make_organism(seed=42)print(f"Compartments: {organism.num_compartments}")print(f"Transport links: {organism.num_transport_links}")for link in organism.transport_links:    print(f"  {link.source} -> {link.target} (mol={link.molecule_id}, rate={link.rate:.4f})")
```

## Simulate

```python
world_timeline = []for _ in range(200):    organism.simulator.step(organism.state)    ws_copy = WorldStateImpl(        organism.state.tree, organism.state.num_molecules,        initial_concentrations=list(organism.state._concentrations),        initial_multiplicities=list(organism.state._multiplicities),    )    world_timeline.append(ws_copy)
```

## Visualize

```python
mol_names = list(chem.molecules.keys())compartment_heatmap(world_timeline, molecule_id=0, title=f"{mol_names[0]} Across Compartments")
```

**Takeaway**: Transport dynamics create spatial gradients across compartments.
