# Demo 02: Equilibrium & Stability**Story**: "How alien biochemistry self-regulates."Run the system at different time horizons and analyze convergence.

```python
import sys, ossys.path.insert(0, os.path.join(os.getcwd(), "demos"))from _shared import make_homeostatic_systemfrom alienbio.bio import check_stabilityfrom alienbio.viz import concentration_trajectory, equilibrium_convergence
```

## Convergence Over Time

```python
for steps in [50, 200, 500]:    s = make_homeostatic_system(seed=42)    tl = s.run(steps)    stab = check_stability(tl, window=min(50, steps//2), threshold=1e-4)    print(f"Steps={steps}: stable={stab.stable}, max_var={stab.max_variance:.6f}")
```

## Full Run

```python
system = make_homeostatic_system(seed=42)timeline = system.run(steps=500)stability = check_stability(timeline, window=100, threshold=1e-4)for mol, var in stability.variance.items():    print(f"{mol}: variance={var:.8f}")
```

```python
concentration_trajectory(timeline, title="Equilibrium Trajectories")
```

```python
equilibrium_convergence(timeline, stability, title="Convergence Analysis")
```

**Takeaway**: Homeostatic systems converge to steady state via balanced source/degradation reactions.
