# Combo C: Ecosystem Under Stress"Survival under pressure."Multi-compartment organism with operating envelope, stressed by toxin spike.

```python
import sys, ossys.path.insert(0, os.path.join(os.getcwd(), "demos"))from _shared import make_organismfrom alienbio.bio import WorldStateImplfrom alienbio.scenarios.organism_features import OperatingEnvelopefrom alienbio.viz import compartment_heatmap, envelope_timeline
```

## Setup

```python
organism, chem = make_organism(seed=42)mol_names = list(chem.molecules.keys())envelope = OperatingEnvelope()for comp_id in range(1, organism.num_compartments):    envelope.add(molecule_id=0, compartment_id=comp_id, low=0.5, high=8.0)print(f"Compartments: {organism.num_compartments}")print(f"Envelope: {mol_names[0]} in [0.5, 8.0]")
```

## Phase 1: Equilibrium

```python
def copy_ws(ws):    return WorldStateImpl(ws.tree, ws.num_molecules,        initial_concentrations=list(ws._concentrations),        initial_multiplicities=list(ws._multiplicities))timeline = []for _ in range(200):    organism.simulator.step(organism.state)    timeline.append(copy_ws(organism.state))violations = sum(1 for ws in timeline[-50:] if not envelope.check(ws).viable)print(f"Violations in last 50 steps: {violations}")
```

## Phase 2: Toxin Spike

```python
current = organism.state.get(1, 0)organism.state.set(1, 0, current + 20.0)print(f"Spiked {mol_names[0]} in compartment 1 by +20.0")
```

## Phase 3: Recovery

```python
for _ in range(300):    organism.simulator.step(organism.state)    timeline.append(copy_ws(organism.state))violations_after = sum(1 for ws in timeline[-50:] if not envelope.check(ws).viable)total = sum(1 for ws in timeline if not envelope.check(ws).viable)print(f"Violations in last 50: {violations_after}")print(f"Total violations: {total}/{len(timeline)}")
```

## Visualize

```python
compartment_heatmap(timeline, molecule_id=0,    compartment_names={i: f"Comp {i}" for i in range(organism.num_compartments)},    title=f"{mol_names[0]} Under Stress")
```
