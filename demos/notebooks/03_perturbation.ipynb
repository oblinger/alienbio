# Demo 03: Perturbation & Recovery**Story**: "What happens when you poke it?"Inject a spike and remove a reaction to see how systems respond.

```python
import sys, ossys.path.insert(0, os.path.join(os.getcwd(), "demos"))from _shared import make_homeostatic_systemfrom alienbio.bio import inject_spikefrom alienbio.viz import perturbation_response, concentration_trajectory
```

## Spike Injection

```python
system = make_homeostatic_system(seed=42)baseline_system = make_homeostatic_system(seed=42)baseline_tl = baseline_system.run(steps=300)system.run(steps=200)result = inject_spike(system, molecule="A", amount=20.0, recovery_steps=300, tolerance=0.15)perturbed_tl = system.run(steps=300)print(f"Recovered: {result.recovered}")print(f"Max deviation: {result.max_deviation:.4f}")print(f"Recovery step: {result.recovery_step}")
```

```python
perturbation_response(baseline_tl, perturbed_tl, result, title="Spike Recovery")
```

## Reaction Removal

```python
system2 = make_homeostatic_system(seed=42)system2.run(steps=200)system2.chemistry.reactions["degrade_a"].set_rate(0.0)drift_tl = system2.run(steps=300)final = drift_tl[-1]for name, _ in final.items():    print(f"{name}: {final[name]:.4f}")
```

```python
concentration_trajectory(drift_tl, title="Reaction Removal Drift")
```

**Takeaway**: Healthy systems recover from spikes but drift from structural damage.
